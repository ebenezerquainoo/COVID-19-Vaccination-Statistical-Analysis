## 2026 Computational Extension

This notebook extends the exploratory analysis by examining the relationship between **sex/gender** and **vaccination status** using a 2×2 contingency table.

Because the unvaccinated group is very small, the analysis uses Fisher's exact test and reports an odds ratio with a confidence interval.

## Research question

**Question:** Is reported vaccination status associated with sex/gender in this survey dataset?

**H₀:** Sex/gender and vaccination status are independent.

**H₁:** Sex/gender and vaccination status are associated.

This is an exploratory analysis of this survey dataset and should not be generalized to the wider UMaT student population.

```
import pandas as pd
from scipy.stats import fisher_exact
from statsmodels.stats.contingency_tables import Table2x2

df = pd.read_csv("../data/project_dataset_cleaned_PRIVATE.csv")
df.head()
```

## 1. Construct the contingency table

Only respondents with both known sex/gender and known vaccination status are included. Missing values are not treated as evidence of either category.

```
analysis_df = df.dropna(subset=["Sex/Gender", "Vaccination_status"])

table = pd.crosstab(
    analysis_df["Sex/Gender"],
    analysis_df["Vaccination_status"]
).reindex(
    index=["Female", "Male"],
    columns=["Vaccinated", "Not vaccinated"],
    fill_value=0
)

display(table)
```

## 2. Fisher's exact test

Fisher's exact test is appropriate for a 2×2 table when the sample information is sparse. It tests the null hypothesis of independence using the observed margins.

```
odds_ratio, p_value = fisher_exact(table.to_numpy())

print(f"Odds ratio: {odds_ratio:.3f}")
print(f"Fisher's exact p-value: {p_value:.3f}")
```

## 3. Odds ratio and 95% confidence interval

The odds ratio provides an effect estimate, while the confidence interval describes its uncertainty.

```
t22 = Table2x2(table.to_numpy())

ci_low, ci_high = t22.oddsratio_confint()

print(f"Odds ratio: {t22.oddsratio:.3f}")
print(f"95% CI: ({ci_low:.3f}, {ci_high:.3f})")
```

## 4. Interpretation

Only two respondents were classified as not vaccinated. Therefore, the relationship between sex/gender and vaccination status is estimated very imprecisely.

A non-significant exact-test result does not prove that sex/gender has no relationship with vaccination status. It means that this dataset does not provide sufficient evidence to reject independence.

The odds ratio should be treated as descriptive and should not be generalized beyond this survey.

## Conclusion

The sex/gender–vaccination relationship is difficult to estimate precisely because almost all respondents were classified as vaccinated. The main lesson is therefore methodological: appropriate test selection and transparent reporting of uncertainty are more important than forcing a strong conclusion from sparse data.
